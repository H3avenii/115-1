```python
身高 = float(input("請輸入身高（公分）："))
體重 = float(input("請輸入體重（公斤）："))

身高公尺 = 身高 / 100
bmi = 體重 / (身高公尺 ** 2)
if bmi < 18.5:
    status = "過輕"
elif 18.5 <= bmi < 24:
    status = "正常"
else:
    status = "過重"

print(f"你的 BMI 值為：{bmi:.2f}，判定為：{status}")
```
