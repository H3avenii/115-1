```python
import random
贏 = False
for day in range(1, 4):
    幸運數字 = random.randint(1, 50)
    猜 = int(input(f"今天第{day}天,請輸入1-50數字"))
    print(f"今天的幸運數字是{幸運數字}")
    if 猜 == 幸運數字:
        贏 = True
        break
    else:
        print("你沒中獎，明天再來\n")
if not 贏:
    print("很遺憾3天都沒中獎，下次再試試吧！")
```
