```python
def f1(p,q):
  return p*q
def f2(p,q):
  total=f1(p,q)
  return total*0.9
p=int(input("請輸入單價： "))
q=int(input("請輸入數量： "))
print("原價金額：", f1(p,q))
print("9折金額：", f2(p,q))
```
