```python
inventory = set()
print("初始背包:", inventory)
inventory.add("劍")
inventory.add("盾")
inventory.add("藥水")
print("撿到道具後的背包:", inventory)
inventory.add("藥水")
print("再次撿到藥水後的背包:", inventory)
inventory.remove("盾")
print("使用盾後的背包:", inventory)
dropped_item = inventory.pop()
print("掉落的道具:", dropped_item)
print("遊戲結束，最後背包:", inventory)
```
