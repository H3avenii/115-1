```python
try:
  account=input("請輸入帳號：")
  password=input("請輸入密碼：")
  if account =="" or password =="":
    raise ValueError("帳號或密碼不可為空")
except ValueError:
  print("帳號或密碼不可為空")
else:
  if account =="admin" and password =="1234":
    print("登入成功")
  else:
    print("登入失敗")
finally:
  print("系統結束")
```
